```python
import os
import whisper

# Paths
audio_dir = "/content/whisper_audio"
output_dir = "/content/transcripts_medium"
os.makedirs(output_dir, exist_ok=True)

# Load Whisper medium model
model = whisper.load_model("medium")

# Transcribe each audio file
for fname in sorted(os.listdir(audio_dir)):
    if fname.endswith(".wav"):
        audio_path = os.path.join(audio_dir, fname)
        output_path = os.path.join(output_dir, f"{os.path.splitext(fname)[0]}.txt")

        print(f"🔍 Transcribing: {fname}")
        result = model.transcribe(audio_path)

        # Save result to its own text file
        with open(output_path, "w") as f:
            f.write(result["text"])

print("✅ All files transcribed and saved individually.")
```
